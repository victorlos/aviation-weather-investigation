Import the required libraries

```python
import os
import snowflake.connector
from dotenv import load_dotenv
```

Load dotenv variables and create a snowflake connection

```python
load_dotenv()  # Carrega variáveis de ambiente do arquivo .env

conn = snowflake.connector.connect(
    user=os.environ["SNOWFLAKE_USER"],
    password=os.environ["SNOWFLAKE_PASSWORD"],
    account=os.environ["SNOWFLAKE_ACCOUNT"],
    warehouse=os.environ["SNOWFLAKE_WAREHOUSE"],
    database=os.environ["SNOWFLAKE_DATABASE"],
    schema=os.environ["SNOWFLAKE_SCHEMA"],
)
```

Run a loop for all files in /setup and run their SQL content in snowflake

```python
for file_name in os.listdir("../dbt-aviation-weather/setup"):
    with open(f"../dbt-aviation-weather/setup/{file_name}", "rb") as f:
        file_content = f.read()

        sql = file_content.decode("utf-8")
        print(f"Executing SQL from {file_name}...")
        cursors = conn.execute_string(sql, remove_comments=True)
        for cursor in cursors:
            print(cursor.fetchall())
            while cursor.nextset():
                print(cursor.fetchall())

conn.close()
```
